```python
df = pd.read_csv("/home/hanysaa/custdata.tsv", sep="\t")
```

```python
df["housing.type"].isna().sum()
```

```python
df_clean = df.dropna(subset=["housing.type"])
```

```python
df_clean.shape
```

```python
df_clean["housing.type"].isna().sum()
```

```python
df_clean["housing.type"].value_counts()
```

```python
housing_counts = df_clean["housing.type"].value_counts().sort_values()
housing_counts
```

```python
import matplotlib.pyplot as plt
```

```python
plt
```

```python
housing_counts.plot(kind="bar")

plt.title("Customers by Housing Type")
plt.xlabel("Housing Type")
plt.ylabel("Number of Customers")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```
