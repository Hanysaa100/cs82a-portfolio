```python
import pandas as pd
```

```python
books = pd.read_csv("allbacks.csv")
books.head()
```

```python
books["cover"].sort_values().reset_index(drop=True)
```

the Median in (pb) which is the value of number 8th

```python
sorted_covers = books["cover"].sort_values().reset_index(drop=True)
median_cover = sorted_covers.iloc[len(sorted_covers) // 2]

print("Median cover type:", median_cover)
```

```python
mean_weight = books["weight"].mean()
print("Mean weight of all books:", mean_weight)
```

The mean weight of all books is 706.67 grams

```python
mean_by_cover = books.groupby("cover")["weight"].mean()
print(mean_by_cover)
```


The mean of hb weighs about 796.43 grams, while the meanof pb weighs about 628.13 grams.

The hardcover books are about 168.30 grams heavier than the softcover, which suggests that the cover type is associated with a difference in book weight in this inventory.

```python
print("Median cover type:", median_cover)
print("Mean weight of all books:", mean_weight)
print("Mean hardcover weight:", mean_by_cover["hb"])
print("Mean paperback weight:", mean_by_cover["pb"])
```
