```python
import pandas as pd

df = pd.read_csv("/home/hanysaa/truck.csv")

df.head()
```

```python
df["X"].corr(df["Y"])
```

```python
import matplotlib.pyplot as plt

plt.scatter(df["X"], df["Y"])
plt.xlabel("List Price (thousands of dollars)")
plt.ylabel("Best Sale Price (thousands of dollars)")
plt.title("Truck List Price vs. Best Sale Price")
plt.show()
```

```python
df["X"].min(), df["X"].max()
```

```python
import numpy as np

slope, intercept = np.polyfit(df["X"], df["Y"], 1)

list_price = 17.40
predicted_best_price = slope * list_price + intercept

predicted_best_price
```

My confidence in the prediction is very high because the Pearson correlation equal 0.9985 which is very close to 1 .which reprsent strong linear relationship between list price and sale price.I choose a list price of 17400 and i got a 15244 as a best price which both results within the range .so the prediction is supported the observation data.however.the sale price is estimated and not the actual sale price.
