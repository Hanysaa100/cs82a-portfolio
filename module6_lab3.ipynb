```python
import sqlite3
import pandas as pd
import matplotlib.pyplot as plt
```

```python
conn = sqlite3.connect("module6_lab3.db")
```

```python
conn.execute("DROP TABLE IF EXISTS customers")

conn.execute("""
CREATE TABLE customers (
    id INTEGER PRIMARY KEY,
    name TEXT,
    city TEXT,
    age INTEGER
)
""")
```

```python
conn.executemany("""
INSERT INTO customers (id, name, city, age)
VALUES (?, ?, ?, ?)
""", [
    (1, "Alex Rivera", "Los Angeles", 25),
    (2, "Maria Chen", "Santa Monica", 31),
    (3, "Daniel Smith", "Los Angeles", 42),
    (4, "Sarah Johnson", "Pasadena", 28),
    (5, "Anthony Brown", "Long Beach", 35),
    (6, "Jennifer Davis", "Los Angeles", 22),
    (7, "Andrew Wilson", "Santa Monica", 39),
    (8, "Emily Garcia", "Pasadena", 27)
])
```

```python
pd.read_sql("SELECT * FROM customers", conn)
```

```python
pd.read_sql("""
SELECT *
FROM customers
WHERE city = 'Los Angeles'
""", conn)
```

```python
pd.read_sql("""
SELECT *
FROM customers
ORDER BY age ASC
""", conn)
```

```python
pd.read_sql("""
SELECT *
FROM customers
WHERE name LIKE '%an%'
""", conn)
```

```python
city_counts = pd.read_sql("""
SELECT city, COUNT(*) AS customer_count
FROM customers
GROUP BY city
""", conn)

city_counts
```

```python
city_counts.plot(
    kind="bar",
    x="city",
    y="customer_count",
    legend=False
)

plt.title("Number of Customers by City")
plt.xlabel("City")
plt.ylabel("Number of Customers")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
