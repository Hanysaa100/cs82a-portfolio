```python
# Convert pounds to kilograms

def pounds_to_kg(pounds):
    kilograms = pounds * 0.453592
    return kilograms
```

```python
# First call
print(pounds_to_kg(100))

# Second call
print(pounds_to_kg(150))

# Third call: replace 170 with your actual weight in pounds
my_weight_pounds = 170
print(pounds_to_kg(my_weight_pounds))
```
