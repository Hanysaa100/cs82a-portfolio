```python
# 1. Import necessary libraries for visualization
import matplotlib.pyplot as plt
import seaborn as sns

# 2. Print the missing-value count for each column
# We use the .isnull().sum() method to find out how many missing (NaN) values are in each column.
print("--- Missing Values Per Column ---")
missing_data = df.isnull().sum()
print(missing_data)
print()

# 3. Calculate and print mean flipper_length_mm by species
# pandas.groupby automatically excludes missing (NaN) values from the calculation,
# ensuring we get an accurate average for each species.
print("--- Mean Flipper Length by Species ---")
mean_flipper = df.groupby("species")["flipper_length_mm"].mean()
print(mean_flipper)
print()
print("Verification Hint: You can verify this by filtering the DataFrame for each species ")
print("individually (e.g., df[df['species'] == 'Adelie']['flipper_length_mm'].mean())")

# 4. Create a labeled seaborn boxplot of flipper_length_mm by species
# We set the overall background style first
sns.set_theme(style="whitegrid")

# Create a figure with custom dimensions (width=8 inches, height=6 inches)
plt.figure(figsize=(8, 6))

# Create the boxplot. We map 'hue' to 'species' and set 'legend=False' to avoid warnings
sns.boxplot(data=df, x="species", y="flipper_length_mm", hue="species", palette="Set2", legend=False)

# 5. Add beginner-friendly labels and a title
plt.title("Comparison of Flipper Length by Penguin Species", fontsize=14, pad=15)
plt.xlabel("Penguin Species", fontsize=12)
plt.ylabel("Flipper Length (mm)", fontsize=12)

# Render the complete chart on the screen
plt.show()
```
